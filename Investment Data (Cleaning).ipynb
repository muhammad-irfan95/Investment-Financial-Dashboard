```python
# 2. Names aur IDs ke liye 'Unknown' fill karein
name_id_cols = ['Participant Name', 'Participant ID', 'Partner Name', 'Partner ID']
for col in name_id_cols:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')

# 3. Categorical aur text columns ke liye 'Not Specified' fill karein
cat_cols = ['Fund Type', 'Status', 'Deposit Branch Name', 'Payment Mode', 'Role']
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].fillna('Not Specified')

# 4. PAN, Bank details aur Dates ke liye 'Not Provided' fill karein
info_cols = ['Participant PAN', 'Partner PAN', 'Participant Bank', 'Participant IFSC', 
             'Partner Bank', 'Partner IFSC', 'Investment Date', 'Payout Date','Partner Phone']
for col in info_cols:
    if col in df.columns:
        df[col] = df[col].fillna('Not Provided')

# 5. Bank Account Numbers ke liye 0 fill karein
ac_cols = ['Participant A/C', 'Partner A/C']
for col in ac_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# 6. Payout Status ke liye 'Pending' fill karein
if 'Payout Status' in df.columns:
    df['Payout Status'] = df['Payout Status'].fillna('Pending')

# 7. Saare Financial aur Numeric columns ke liye 0 (zero) fill karein
numeric_cols = [
    'Capital (₹)', 'Total Profit (₹)', 'Net Payable (₹)', 'ROI (%)', 
    'Duration (mo)', 'Paid (mo)', 'Remaining (mo)', 'Monthly Payout (₹)', 
    'First Month Payout (₹)', 'Final Month Payout (₹)', 'Payout This Month (₹)', 
    'Comm Rate (%)', 'Partner Comm This Month (₹)', 'Partner Monthly Base (₹)', 
    'Partner Lifetime (₹)'
]
for col in numeric_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# Check karne ke liye ki sabhi null values khatam hui ya nahi
print(df.isnull().sum())
```

```python
df.head()
```

```python
# Sirf Participant ke liye 'Unknown' rakhein, Partner ke liye nahi
name_id_cols = ['Participant Name', 'Participant ID']
for col in name_id_cols:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')

# Partner IDs ke liye blank/NaN hi rehne dein taaki unwanted rows na judein
partner_cols = ['Partner Name', 'Partner ID']
for col in partner_cols:
    if col in df.columns:
        df[col] = df[col].fillna(pd.NA)
```
